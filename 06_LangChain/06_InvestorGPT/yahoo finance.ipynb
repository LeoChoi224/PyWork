```python
records = []

for year, row in df_t.iterrows():
    # print(year, type(row))
    record = {"fiscalDateEnding": year}  # 회계연도 마감일
    for key, value in row.items():
        # print('\t', key, value)

        if value is None or (isinstance(value, float) and value != value):  # value != value <- NaN 여부 체크
            record[str(key)] = None
        else:
            record[str(key)] = value

    records.append(record)

records
```

# 주가 정보

```python
# 최근 6개월, weekly 주가 데이터 (주봉 데이터)
hist = ticker.history(period='6mo', interval='1wk', auto_adjust=False)
hist

# auto_adjust=False는 주식 데이터의 과거 가격(시가, 고가, 저가, 종가)을 가져올 때, 
# 배당(Dividends)이나 액면분할(Stock Splits) 같은 기업 이벤트를 반영하여 과거 가격을 자동으로 수정하지 않겠다는 의미
```

```python
# 최근 12주만 가져오기
hist = hist.tail(12)
hist
```

```python
"""
[{'date': '2026-03-02',
  'open': 390.6,
  'high': 408.62,
  'low': 385.39,
  'close': 396.73,
  'volume': 301991600},
 {'date': '2026-03-09',
  'open': 390.05,
  'high': 416.38,
  ...
"""
None
```

```python
weekly_bars = []
for ts, row in hist.iterrows():
    weekly_bars.append({
        "date": ts.strftime("%Y-%m-%d"),  # JSON 변환을 위해
        "open": round(float(row["Open"]), 4),
        "high": round(float(row["High"]), 4),
        "low": round(float(row["Low"]), 4),
        "close": round(float(row["Close"]), 4),
        "volume": int(row["Volume"]),
    })

weekly_bars    
```

```python
# 간단한 성과  요약
first_close = weekly_bars[0]["close"]
last_close = weekly_bars[-1]["close"]
change_pct = ((last_close - first_close) / first_close) * 100.0
```

```python
result = {
    "symbol": symbol,
    "interval": "1 week",
    "weeks_returned": len(weekly_bars),
    "summary": {
        "first_close": first_close,
        "last_close": last_close,
        "period_change_pct": round(change_pct, 2),
        "period_high": round(float(hist["High"].max()), 4),
        "period_low": round(float(hist["Low"].min()), 4),
    },
    "weekly_bars": weekly_bars,
}

result
```

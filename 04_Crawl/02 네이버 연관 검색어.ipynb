# 네이버 연관 검색어

- 동적으로 요청하는 URL로 요청하기

- **JSONP(JSON with Padding)** 는
웹 브라우저의 동일 출처 정책(Same-Origin Policy) 때문에 AJAX 요청으로 다른 도메인(domain) 의 데이터를 직접 가져올 수 없을 때 사용하던 우회 방식입니다.


```python
import requests
import json
import re
```

```python
url = f"https://ac.search.naver.com/nx/ac?q=%EC%A0%80&con=1&frm=nv&ans=2&r_format=json&r_enc=UTF-8&r_unicode=0&t_koreng=1&run=2&rev=4&q_enc=UTF-8&st=100&ackey=sinrtmq6&_callback=_jsonp_4"

response = requests.get(url)
response
```

```python
response_text = response.text
response_text
```

```python
response_text = response_text.replace('\n', '')
response_text
```

```python
# _jsonp_#(....) 괄호 안쪽의 ... 내용만 추출

match = re.search(r'\((.*)\)', response_text)
match
```

```python
if match:
    json_str = match.group(1)
    # print(json_str)
    data = json.loads(json_str)
    print(data)
```

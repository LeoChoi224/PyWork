# 종합: 함수로 정의

```python
def crawl_yes24(query, page=1):
    print(f'🔶 {query} 검색 page{page}...')

    url = f'https://www.yes24.com/product/search?domain=BOOK&query={query}&page={page}&size=24'

    response = requests.get(url)
    if response.status_code != 200: return

    dom = BeautifulSoup(response.text, 'html.parser')
    book_list = dom.select('#yesSchList > li')

    result = []

    for book_item in book_list:
        # 썸네일
        img_src = book_item.select_one("div.item_img > div.img_canvas > span > span > a > em > img").attrs['data-original'].strip()
    
        # 책제목
        book_title = book_item.select_one("div.item_info a.gd_name").text.strip()
    
        # 책URL
        book_url = 'https://www.yes24.com' + book_item.select_one("div.item_info a.gd_name").attrs['href'].strip()
    
        # 가격
        price = book_item.select_one("div.item_info > div.info_row.info_price > strong > em").text.strip()
        price = int(re.sub('[^0-9]', '', price))

        result.append({
            'thumbnail_url': img_src,
            'book_title': book_title,
            'book_url': book_url,
            "price": price,
        })

    return result
```

```python
result = crawl_yes24('파이썬', 3)
result
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(result)
df
```

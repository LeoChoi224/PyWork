### 문제 5. 범주형 대상 Class 컬럼을 수치로 바꾸어 표현하기

#### 5-1 Class_value 컬럼 추가하여 수치로 표현하기

```python
# Class_value 란 컬럼을 새로이 추가하여
# Class 의 L, M, H를 숫자로 바꾸어 표현하기 (eg. L: -1, M: 0, H:1)


```

```python
df['Class_value'] = df['Class'].apply(lambda x: {'L': -1, 'M': 0, 'H': 1}[x])

df[['Class', 'Class_value']].head()
```

```python
df['Class_value'].value_counts()
```

#### 5-2 학습주제(topic)별 Class_value 평균값 구하고 시각화 하기

```python
# 숫자로 바꾼 Class_value 컬럼을 이용하여
# 학습주제(topic) 별 Class_value 의 평균값을 구한뒤 
# 내림차순으로 정렬된 결과를 시각화하기 (bar plot 으로 그리세요)
```

```python
# TODO
topic_mean = df.groupby('Topic')['Class_value'].mean().sort_values(ascending=False)
topic_mean
```

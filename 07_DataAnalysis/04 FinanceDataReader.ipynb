## Overview

FinanceDataReader 는 과거의 pandas-datareader 를 대체하기 위해 제작됨.  주요 특징은 아래와 같다

<ul>
    <li>Stock price(Word wide): AAPL, AMZN, GOOG ...   (종목별 주가: 세계)</li>
    <li>Stock price(KRX): 005930(Samsung), 091990(Celltrion Healthcare) ... (종목별주가: 한국)</li>
    <li>Indexes: KOSPI, KOSDAQ, DJI, IXIC, US500(S&P 500) ... (주요 지수들)</li>
    <li>Exchanges: USD/KRX, USD/EUR, CNY/KRW ... (환율)</li>
    <li>Cryptocurrency: BTC/USD (Bitfinex), BTC/KRW (Bithumb)  (가상화폐)</li>
    <li>Symbol listings: KRX (KOSPI, KODAQ, KONEX), NASDAQ, NYSE, AMEX and S&P 500 ( ???? )</li>
</ul>


```python
import FinanceDataReader as fdr
```

```python
df = fdr.DataReader('AAPL', '2026-08-01')
df
```

```python
df = fdr.DataReader('AMZN', '2026-07-02', '2026-07-31')
df
```

```python
df = fdr.DataReader('005930', '2026-07-01', '2026-08-01')
df
```

```python
# df = fdr.DataReader('005930', '2026-07-01', '2026-08-01', exchange="KRX")
df = fdr.DataReader('KRX:005930', '2026-07-01', '2026-08-01')
df
```

```python
# '000150': 두산
df = fdr.DataReader('000150', '2026-07-01', '2026-08-01')
df
```

```python
df = fdr.DataReader('000150', '2026-07-01', '2026-08-01', exchange='SSE')
df
```

```python
# KOSPI index
df = fdr.DataReader('KS11', '2026-07-01', '2026-07-31')
df
```

```python
df = fdr.DataReader('USD/KRW', '2026-07-01', '2026-07-31')
df
```

```python
# symbol list
df = fdr.StockListing('KRX')
df
```

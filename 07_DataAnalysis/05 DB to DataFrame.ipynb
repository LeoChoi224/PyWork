# 특정 DB의 내용을 CSV로 저장

```python
dbname = 'sakila'

engine = create_engine(f"mysql+pymysql://root:1234@localhost/{dbname}?charset=utf8")

dataset_path = r'/Users/leo/LeoData/MyCode/KoreaITAcademy/LangChain/dataset'

path = os.path.join(dataset_path, 'mysql', dbname)
if not os.path.exists(path):
    print(path, '를 생성합니다')
    os.makedirs(path)

table_names = pd.read_sql('SHOW TABLES', engine).iloc[:, 0]    

for table_name in table_names:
    query = f"SELECT * FROM {table_name}"
    table_df = pd.read_sql(query, engine)

    # DB -> DataFrame -> CSV 저장
    csv_path = os.path.join(dataset_path, 'mysql', dbname, f'{table_name}.csv')

    table_df.to_csv(csv_path, index=False, encoding='utf8')
    print(f'{csv_path} 생성')

engine.dispose()  # 연결 닫기
    
```

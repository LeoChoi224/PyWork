# AI Agent
AI 에이전트(AI Agent)는 주어진 목표를 달성하기 위해 주변 환경을 인지하고, 스스로 판단하여 행동을 수행하는 자율적인 인공지능 시스템을 말합니다.

단순히 사용자의 질문에 답을 생성하는 '챗봇(Chatbot)'이나 일정한 규칙에 따라 동작하는 '자동화 프로그램'을 넘어, **목표를 달성하기 위한 최선의 경로를 스스로 계획하고 실행**한다는 점에서 차세대 AI 기술로 주목받고 있습니다.

AI 에이전트의 핵심 메커니즘을 쉽게 이해하기 위해 구조를 살펴보면 다음과 같습니다.

---

## 1. AI 에이전트의 4대 핵심 요소

AI 에이전트가 자율적으로 작동할 수 있는 것은 대형언어모델(LLM)을 '두뇌'로 삼고 아래의 4가지 핵심 능력을 결합했기 때문입니다.

* **목표 및 계획 수립 (Planning):** 거대한 목표가 주어지면 이를 해결 가능한 작은 단위의 작업(Sub-tasks)으로 쪼갭니다. 과거의 실수를 돌아보고 계획을 수정하는 '자기 반성(Self-reflection)' 능력도 포함됩니다.
* **기억장치 (Memory):**
* *단기 기억:* 현재 진행 중인 대화나 작업 맥락을 기억합니다.
* *장기 기억:* 외부 데이터베이스나 벡터 DB를 활용해 오랜 시간이 지난 후에도 필요한 정보를 찾아내고 유지합니다.


* **도구 활용 (Tools):** AI 자체의 한계를 극복하기 위해 외부 도구를 사용합니다. 계산기 활용, 웹 브라우징, API 호출, 데이터베이스 조회 등이 이에 해당합니다.
* **행동 (Action):** 계획과 도구를 바탕으로 이메일 발송, 결제, 코드 실행 등 실제 환경이나 시스템에 변화를 주는 결과물을 만들어냅니다.

---

## 2. 챗봇(Chatbot) vs AI 에이전트(Agent)

| 구분 | 일반 AI 챗봇 (예: 기본 ChatGPT, Gemini) | AI 에이전트 (AI Agent) |
| --- | --- | --- |
| **작동 방식** | 사용자의 질문(프롬프트)에 수동적으로 답변 | 목표를 주면 스스로 판단하여 능동적으로 행동 |
| **수행 단계** | 1문 1답 형태의 단발성 상호작용 | 목표 달성을 위해 멀티 스텝(Multi-step) 작업 수행 |
| **도구 활용** | 텍스트 생성 및 내장된 기능 위주 활용 | 브라우저 제어, API 연동 등 외부 도구 자율 제어 |
| **예시** | "서울 맛집 3곳 추천해줘" $\rightarrow$ 목록 작성 | "내 일정과 예산에 맞춰 서울 여행 계획을 짜고 숙소까지 예약해줘" $\rightarrow$ 대리 수행 |

---

## 3. 대표적인 활용 시나리오

AI 에이전트는 이미 다양한 영역에서 실험 단계를 넘어 실무에 적용되고 있습니다.

* **업무 자동화 및 비서:** 사용자의 메일함을 관리하고, 중요도를 판단해 답장 초안을 작성하거나, 회의 일정을 조율하고 캘린더에 자동으로 등록합니다.
* **웹 브라우저 제어 및 예약:** 인간 대신 웹사이트에 접속해 로그인하고, 원하는 조건의 항공권이나 식당을 찾아 최종 결제 직전 단계까지 자율적으로 진행합니다.
* **소프트웨어 개발 (AI 개발자):** 코드의 버그를 찾아달라고 요청하면, 스스로 전체 소스코드를 분석하고 수정한 뒤 테스트까지 완료하여 패치 코드를 제안합니다.
* **디바이스 제어 (스크린 자동화):** 스마트폰이나 PC의 화면(UI)을 직접 인식하고 터치/클릭하여 배달 앱에서 음식을 주문하거나 택시를 호출하는 등의 작업을 수행합니다.

> 💡 **요약하자면**
> 기존의 AI가 질문에 답을 주는 **'지식 가이드'** 였다면, AI 에이전트는 사용자를 대신해 복잡한 업무를 끝마쳐주는 **'자율적인 대리인(Agent)'** 이라고 볼 수 있습니다.

# 1.그래프(Graph) 자료구조의 이해

랭그래프를 제대로 이해하려면 먼저 그래프 Graph 자료구조에 대한 기본 개념을 알아야 합니다. 

## 노드 & 에지

그래프(Graph)는 **노드(Node)** 와 **에지(Edge)** 로 구성된 자료구조로, 객체 간의 **관계**를 표현하는 데 매우 효과적입니다. 

※자료구조에선 노드를 버텍스(Vertex), 에지를 아크(Arc) 라고도 표기하기도 함.

![](https://www.dropbox.com/scl/fi/texuh7nrxuv38s241xixd/13817670-figure-1.png?rlkey=3x08ugk6ub82eohdkpz3sbqf9&st=xnaacb9s&dl=1)


![](https://www.dropbox.com/scl/fi/2nrsxn30ipjd4to94y3md/graph-vertex-edge.png?rlkey=gw1hbhtrisk0t721e2yx6h5ol&st=trjc93un&dl=1)

## Directed vs. Undirected
그래프는 방향성에 따라 방향 그래프 Directed Graph 와 무방향 그래프 Undirected Graph 로 나뉩니다. 

![](https://www.dropbox.com/scl/fi/y1wirvs106x2xco9cbp23/graph-direction.png?rlkey=4apvuf3wwcuzzkg23x5ydssqr&st=fsyytskc&dl=1)

**랭그래프는 방향 그래프를 기반**으로 합니다. 이는 작업의 흐름이 특정 방향을 가지기 때문입니다. 
예를 들어 A 작업이 완료된 후 B 작업이 실행되는 것을 그래프로 명확하게 표현할 수 있습니다.

## Cyclic vs. Acyclic 
![](https://www.dropbox.com/scl/fi/ks6p5dq82k8k2g8fljq6x/2z9J2E5.png?rlkey=3ol814oehfu49p8kkgnrb15jc&st=vs5ida6l&dl=1)
그래프의 또 다른 중요한 특성은 순환(cycle)의 존재 여부입니다. 
순환이 없는 그래프를 **DAG(Directed Acyclic Graph)** 라고 하며, 많은 워크플로 시스템이 DAG만을 이용합니다. 

랭그래프는 __순환이 있는 그래프도 사용할 수 있습니다__. 이를 통해 조건이 만족될 때까지 특정 작업을 반복하거나, 
이전 단계로 돌아가는 등의 복잡한 패턴을 구현할 수 있습니다.

# 2.랭그래프의 핵심 개념
랭그래프에서 가장 중요한 개념은 **상태(state), 노드(node), 에지(edge)** 다

## 상태(state)
**상태**는 그래프 실행 과정에서 지속적으로 유지되는 데이터로 각 노드가 실행될 때마다 읽고 쓸 수 있습니다. 

이는 단순한 변수 전달을 넘어 전체 워크플로의 컨텍스트를 관리하는 **중앙 저장소 역할**을 합니다. 

상태는 일반적으로 **TypedDict**나 **Pydantic 모델**로 정의되어 타입 안정성을 보장합니다.

![](https://www.dropbox.com/scl/fi/xde0lswqc2uuzolfwtn6u/66dba5f465f6e9a2482ad935_simple-graph1.png?rlkey=actdnhup4h33lf924588alp27&st=aqgh7r7q&dl=1)

## 노드(node)
'노드' 는 그래프의 기본 실행 단위입니다.  각 노드는 특정 작업을 수행하는 **함수**나 **에이전트**를 나타내며, 다음과 같은 특징을 가집니다.

- 현재 상태를 **입력**으로 받음
- 특정 **작업을 수행**(LLM 호출, 데이터 처리, 외부 API 호출 등)
- **업데이트된 상태를 반환(리턴)**

노드는 순수함수일 수도 있고, 부수 효과side Effects를 가질 수도 있습니다. 예를 들어 DB에 쓰기 작업을 하거나 외부 API를 호출할 수도 있습니다.

## 에지(edge)
'에지'는 노드 간의 연결을 정의하며, 실행 흐름을 제어합니다. 랭그래프는 두 가지 유형의 에지를 지원합니다.

1. **일반 에지** : 항상 같은 경로로 진행

2. **조건부 에지**: 상태에 따라 다른 노드로 분기

조건부 에지는 동적 라우팅을 가능하게 하여, 런타임에 실행 경로를 결정할 수 있습니다. 
이를 사용하여 AI 에이전트가 상황에 따라 다른 전략을 선택하게 하거나, 오류 처리 로직을 구현하는 데 유용합니다.

* 일반에지는 실선으로 표현하고, 조건부 에지는 점선으로 표기하기도 한다

# 3.헬로 랭그래프 만들기

````python
# 다음과 같은 흐름으로 구성된 랭그래프 예제를 만들어봅시다.

# [시작] --> [Greeting] --> [Processing] --> [종료]
````

````python
from tkinter import Image
from typing import Dict, Any
from pydantic import BaseModel, Field
from IPython.display import Image, display

from langgraph.graph import StateGraph, START, END
````

## 상태정의

````python
# ① 워크플로우 단계 정의
#   워크플로 단계를 정합니다. 이번 예제에서는 인사말을 생성하는 GREETING 단계와 생성된 인사말을 처리하고 
#   메시지를 생성하는 PROCESSING 두 단계로 이루어집니다. enum 혹은 문자열을 사용해도 됩니다. 
#   가독성 측면에서는 예제에서처럼 클래스의 멤버 변수로 넣는 것이 깔끔합니다.

class WorkflowStep:
    GREETING = 'GREETING'      # 인사말을 생성하는 단계
    PROCESSING = 'PROCESSING'  # 생성된 인사말을 처리하고 메세지를 생성하는 단계

# ② 그래프 상태 정의 : 
# State 즉 그래프의 상태는 각 노드마다 참조가 가능한 전체 워크플로 상의 컨텍스트라고 볼 수 있습니다. 
# 상태에는 이름(name), 인사말(greeting), 최종 메시지(processed_message) 세 가지 값이 있습니다.

class GraphState(BaseModel):
    name: str = Field(default="", description="사용자 이름")
    greeting: str = Field(default="", description="생성된 인사말")
    processed_message: str = Field(default="", description="처리된 최종 메세지")
````

## 노드 정의

````python
# ③ 첫번째 노드 함수
# 이번 예제에서는 랭그래프의 기능을 테스트하기 위하여 LLM은 배제하고 코드를 작성했습니다. 
#  generate_greeting()은 일반 함수이며 state 객체를 받아서 인사말 greeting 을 생성하고 반환합니다. 
#  greeting값을 반환하여 State의 greeting값만 업데이트됩니다. <- 랭그래프 실행하게되면 자동적으로 동작함.

def generate_greeting(state: GraphState) -> Dict[str, Any]:
    name = state.name or "정희준 The 군포프린스"
    greeting = f"안녕하세요, {name}님!"
    print(f"🌐[generate_greeting] 인사말 생성: {greeting}")

    return {"greeting": greeting}  # <- state 에서 "greeting" 값을 업데이트 하게 된다.
````

````python
# ④두 번째 노드: 인사말을 처리하고 최종 메시지 생성
#  process_message() 함수는 두 번째 노드의 역할입니다. State에 있는 greeting을 확인하여 최종 메시지를 생성합니다. 
# 그리고 processed_message를 반환하여 해당 값을 업데이트할 수 있게 합니다.

def process_message(state: GraphState) -> Dict[str, Any]:
    greeting = state.greeting
    processed_message = f"{greeting} LangGraph 에 오신 것을 환영합니다!"
    print(f"🌐[process_message] 최종메세지: {processed_message}")

    return {"processed_message": processed_message}
````

## 그래프 정의

````python
# ⑤ 그래프 생성

def create_hello_graph():
    # '상태'를 가진 워크플로(그래프) 객체 생성
    # StateGraph(상태)
    workflow = StateGraph(GraphState)

    # 워크플로 객체에 노드(들) 추가
    # add_node(노드이름, 노드함수)
    workflow.add_node(WorkflowStep.GREETING, generate_greeting)
    workflow.add_node(WorkflowStep.PROCESSING, process_message)
    
    # 엣지 추가 (노드 간 연결)
    # add_edge(노드1, 노드2) 노드1 → 노드2 엣지 형성
    workflow.add_edge(START, WorkflowStep.GREETING)  # 시작 설정 START
    workflow.add_edge(WorkflowStep.GREETING, WorkflowStep.PROCESSING)
    workflow.add_edge(WorkflowStep.PROCESSING, END)   # 그래프 종료 설정 END

    # 그래프 컴파일 -> CompiledStateGraph 객체.  Runnable 객체.  invoke() 호출 가능객체
    app = workflow.compile()

    return app
````

## 그래프 생성및 실행

````python
def main():
    """그래프 생성 및 실행"""
    print("=== Hello 랭그래프 ===\n")
    # 그래프 생성. CompiledStateGraph객체
    app = create_hello_graph()
    
    # 초기 State 준비
    initial_state = GraphState(name="홍무기", greeting="", processed_message="")
    print("🟡 초기상태:", initial_state.model_dump())

    # 그래프 실행
    print("\n🟩 그래프 실행 시작")
    final_state = app.invoke(initial_state)  # 초기상태 전달하여 그래프 호출.  리턴값은 상태의 dict
    print("🟩 그래프 실행 종료\n")
    print('🟡 최종상태:', final_state, type(final_state))
    print(f"\n결과 메세지: {final_state['processed_message']}")

    # 그래프 이미지 생성
    result = app.get_graph().draw_mermaid_png()
    display(Image(result))

    with open("./01_hellograph.png", 'wb') as f:
        f.write(result)

if __name__ == "__main__":
    main()
````

# 참고: CompiledStateGraph 로 실행 호출가능한 메소드들

`CompiledStateGraph`는 LangGraph 내부 `Pregel` 클래스를 상속하면서 LangChain의 `Runnable` 인터페이스도 구현하고 있어서, 실행 관련 메서드가 꽤 많습니다. 역할별로 정리해드릴게요.

## 1. 기본 실행 (한 번에 결과 받기)

| 메서드 | 설명 |
|---|---|
| `invoke(input, config=None)` | 그래프를 동기로 끝까지 실행하고 최종 결과만 반환 |
| `ainvoke(input, config=None)` | 위와 동일하지만 비동기. FastAPI 라우트에서는 이걸 써야 함 |

## 2. 스트리밍 (중간 과정을 흘려받기)

| 메서드 | 설명 |
|---|---|
| `stream(input, config=None, stream_mode=...)` | 동기 스트리밍. `stream_mode`로 `"values"`(매 스텝 전체 상태), `"updates"`(스텝별 변경분만), `"messages"`(LLM 토큰), `"debug"` 등을 선택 |
| `astream(input, config=None, stream_mode=...)` | 위의 비동기 버전 |
| `astream_events(input, config=None, version="v2")` | 그래프 내부의 모든 콜백 이벤트(`on_chat_model_stream`, `on_tool_start` 등)를 잘게 쪼개서 흘려줌. `Runnable` 인터페이스에서 옴 |

`stream_mode`와 `astream_events`의 차이가 헷갈릴 수 있는데: `stream()/astream()`은 **그래프 자체의 노드/상태 변화**를 기준으로 흘려주고, `astream_events()`는 **그 노드 안에서 호출되는 모든 하위 Runnable(LLM, 도구 등)의 콜백 단위**까지 더 세밀하게 흘려줍니다. 토큰 단위 텍스트 스트리밍이 필요하면 결국 `astream_events`(또는 `stream_mode="messages"`)를 써야 해요.

## 3. 배치 실행

| 메서드 | 설명 |
|---|---|
| `batch(inputs, config=None)` | 여러 입력을 한 번에 동기 실행 |
| `abatch(inputs, config=None)` | 비동기 버전 |

## 4. 상태(State) 조회/조작 — `checkpointer`가 있어야 의미 있음

코드에서 `InMemorySaver()`를 `checkpointer`로 넣어주셨기 때문에 이 메서드들이 실제로 동작합니다:

| 메서드 | 설명 |
|---|---|
| `get_state(config)` / `aget_state(config)` | 특정 `thread_id`의 현재 상태 스냅샷 조회 (지금까지의 대화 메시지 등) |
| `get_state_history(config)` / `aget_state_history(config)` | 그 스레드의 상태 변화 이력 전체(체크포인트마다) 조회 — 특정 시점으로 되돌리기(time travel)에 활용 |
| `update_state(config, values, as_node=None)` / `aupdate_state(...)` | 그래프를 실행하지 않고 상태를 직접 수정 (예: 사람이 메시지를 강제로 끼워넣기) |
| `bulk_update_state(...)` / `abulk_update_state(...)` | 여러 상태 업데이트를 한 번에 적용 |

예를 들어 `/chat` 핸들러 옆에 "대화 이력 조회" API를 추가하고 싶다면:

```python
@app.get("/history/{session_id}")
async def get_history(request: Request, session_id: str):
    agent_executor = request.app.state.agent_executor
    config = {"configurable": {"thread_id": session_id}}
    snapshot = await agent_executor.aget_state(config)
    return [m.content for m in snapshot.values["messages"]]
```

이런 식으로 `aget_state`를 바로 활용할 수 있어요.

## 5. 그래프 구조 조회

| 메서드 | 설명 |
|---|---|
| `get_graph(config=None)` / `aget_graph(...)` | 그래프의 노드/엣지 구조를 객체로 반환. `.draw_mermaid()`나 `.draw_mermaid_png()`로 시각화 가능 — 디버깅/문서화용 |
| `get_subgraphs(...)` / `get_subgraphs_async(...)` | 그래프 안에 중첩된 서브그래프 조회 (멀티 에이전트 구조일 때) |

## 6. 기타 (Runnable 공통 메서드)

`with_config(config)`처럼 설정을 미리 바인딩해서 새 인스턴스를 만드는 메서드 등 `Runnable` 인터페이스에서 상속되는 메서드들도 있는데, 일반적인 챗봇 서버 구현에서는 거의 안 씁니다.

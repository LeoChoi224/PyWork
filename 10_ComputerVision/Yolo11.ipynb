# YOLO(You Only Look Once) 모델이란?
- YOLO는 You Only Look Once의 이러한 약자로서 2015년 Joseph Redmon 이 아키텍처를 논문과 함께 발표


- **YOLO**는 이미지 안에서 객체가 "어디에" 있고 "무엇인지"를 한 번의 신경망 연산(single pass)으로 동시에 예측하는 **실시간 객체 탐지(Object Detection)** 모델입니다.

### 핵심 아이디어

- 기존의 R-CNN 계열 방식은

  - ① 후보 영역(Region Proposal)을 먼저 찾고 →

  - ② 각 영역을 분류하는 2단계(two-stage) 방식이라 느렸습니다.

- 반면 YOLO는
  - 이미지를 그리드(grid)로 나눈 뒤,
  - 각 그리드 셀에서 바운딩 박스 좌표와 클래스 확률을 **한 번에** 예측합니다.
  - 이름 그대로 "한 번만 보고(You Only Look Once)" 탐지가 끝나기 때문에 속도가 압도적으로 빠릅니다.

### 출력 구성
- **바운딩 박스(Bounding Box)**: 객체 위치 (x, y, width, height)
- **신뢰도 점수(Confidence Score)**: 박스 안에 객체가 있을 확률
- **클래스 확률(Class Probability)**: 어떤 종류의 객체인지

### 버전 흐름
- **YOLOv1~v3** (2016~2018): Joseph Redmon이 개발한 초기 버전, 원조 개념 정립
- **YOLOv4~v8**: 커뮤니티/Ultralytics 등이 이어받아 속도·정확도 개선, 특히 v8부터 Ultralytics가 `pip install ultralytics`로 학습·추론을 매우 쉽게 만듦
- **YOLOv9~v11**: 아키텍처 경량화, 앵커프리(anchor-free) 구조 도입
- **YOLO26** : 2026년 들어 Ultralytics는 **YOLO26** 을 출시하며 이 흐름을 새롭게 재정의했습니다. YOLO26은 NMS(비최대억제) 없는 종단간 추론과 최적화된 엣지 배포가 특징이며, 다양한 하드웨어 플랫폼에 쉽게 적응 가능한 단순화된 설계를 갖췄습니다. 안정적인 프로덕션 환경에는 현재 YOLO26과 YOLO11 모두 권장됩니다.

- YOLOv4 까지는 C언어 기반의 Darknet 아키텍처가 사용되었지만, YOLOv5 부터는 PyTorch 로 개발이 이루어짐

  1. 빠른속도
  2. 사용자 친화적인 파이썬 인터페이스를 제공하는 YOLO11 (SOTA) 아키텍처로 발전함

### 대표 활용 분야
- 실시간 CCTV/보안 감시
- 자율주행 (보행자·차량 탐지)
- 제조업 결함 검사
- 스포츠 분석, 리테일(재고 트래킹) 등

# Results 객체

일반적으로 `ultralytics` 라이브러리에서 `model(source=...)` 호출의 결과인 `results` 변수는 리스트 형태이며, 각 요소는 `ultralytics.engine.results.Results` 객체입니다. 이 객체는 이미지에서 탐지된 객체에 대한 자세한 정보를 담고 있습니다.

현재 커널 상태에 있는 `results` (detection 결과)는 다음과 같은 주요 속성을 가집니다:

*   **`boxes`**: 탐지된 객체들의 바운딩 박스 정보(좌표, 너비, 높이 등)와 클래스 예측(class prediction), 신뢰도(confidence score)를 포함합니다. `result.boxes.cls.cpu().numpy()`를 통해 클래스 번호를 넘파이 배열로 얻을 수 있습니다.
*   **`masks`**: (세그멘테이션 모델의 경우) 탐지된 각 객체의 마스크 정보(픽셀 단위의 영역)를 포함합니다.
*   **`names`**: 모델이 탐지할 수 있는 클래스 이름(예: 'person', 'car', 'Paper', 'Rock' 등)이 매핑된 딕셔너리입니다. `model.names[class_num]` 형태로 클래스 번호에 해당하는 이름을 가져올 수 있습니다.

## `result.boxes` 객체 설명

`result.boxes` 객체는 `ultralytics.engine.results.Boxes` 타입이며, 이미지 내에서 탐지된 객체들의 바운딩 박스(bounding box) 정보와 관련된 속성들을 포함하고 있습니다. 이 객체는 각 탐지된 객체에 대한 위치, 크기, 클래스, 신뢰도 등의 상세 정보를 제공합니다.

현재 커널 상태에서 `result.boxes`의 출력에 따르면, 다음과 같은 주요 속성들을 가지고 있습니다:

*   **`cls`**: 탐지된 객체들의 클래스 번호(class index)를 나타내는 Torch 텐서입니다. `tensor([63., 64., 66., 67.], device='cuda:0')`는 4개의 객체가 각각 클래스 63, 64, 66, 67로 탐지되었음을 의미합니다.
*   **`conf`**: 각 객체 탐지의 신뢰도(confidence score)를 나타내는 Torch 텐서입니다. 이 값은 모델이 해당 객체를 특정 클래스로 얼마나 확신하는지를 나타냅니다 (예: `0.9576`은 95.76%의 신뢰도).
*   **`data`**: `[x_min, y_min, x_max, y_max, confidence, class_id]` 형식으로 모든 탐지된 객체의 바운딩 박스 좌표, 신뢰도, 클래스 ID를 포함하는 Torch 텐서입니다.
*   **`orig_shape`**: 원본 이미지의 크기를 `(height, width)` 형식으로 나타냅니다 (예: `(516, 640)`).
*   **`xywh`**: 바운딩 박스를 `[x_center, y_center, width, height]` 형식으로 나타내는 Torch 텐서입니다.
*   **`xywhn`**: `xywh`와 같지만, 이미지 크기에 대해 정규화(normalized)된 값입니다 (0에서 1 사이의 값).
*   **`xyxy`**: 바운딩 박스를 `[x_min, y_min, x_max, y_max]` 형식으로 나타내는 Torch 텐서입니다. 이는 일반적으로 이미지의 좌상단을 (0,0)으로 하는 픽셀 좌표입니다.
*   **`xyxyn`**: `xyxy`와 같지만, 이미지 크기에 대해 정규화된 값입니다.

이러한 속성들을 통해 탐지된 각 객체의 위치와 종류, 그리고 그 예측의 정확성을 정량적으로 파악할 수 있습니다.

```python
# Results 객체에서 detect 된 정보 출력
import numpy as np

for result in results:   # result: Results객체
  # 어떤 객체를 몇개 찾았나?
  uniq, cnt = np.unique(
      result.boxes.cls.cpu().numpy().astype(np.int32),  # 검출된 객체들의 class id들
      return_counts=True,
  )
  uniq_cnt_dict = dict(zip(uniq.tolist(), cnt.tolist()))
  print('\n{class id: counts} =', uniq_cnt_dict, '\n')

  # 찾아낸 객체의 클래스이름과 바운딩 박스 좌표
  for idx, box in enumerate(result.boxes):
    print('class num =', int(box.cls),
          ', class_name =', model.names[int(box.cls)],
          ', xyxy=', box.xyxy.cpu().numpy()
          )
```

## YOLO 모델에서의 이미지 세그멘테이션(Image Segmentation)

YOLO(You Only Look Once)는 원래 객체 탐지(Object Detection)를 위한 모델이었지만, 최신 버전(YOLOv5, YOLOv8, YOLO11 등)에서는 **인스턴스 세그멘테이션(Instance Segmentation)** 기능도 지원합니다.

## 객체 탐지 vs 세그멘테이션의 차이

| 구분 | 출력 결과 |
|---|---|
| **객체 탐지(Detection)** | 바운딩 박스(사각형)로 객체 위치만 표시 |
| **이미지 세그멘테이션** | 객체의 정확한 픽셀 단위 윤곽(마스크)까지 표시 |

즉, 탐지는 "이 사각형 안에 고양이가 있다"고 알려주지만, 세그멘테이션은 "이 픽셀들이 정확히 고양이의 형태다"까지 알려줍니다.

## 세그멘테이션의 종류

1. **Semantic Segmentation (시맨틱 세그멘테이션)**
   - 픽셀마다 클래스를 분류 (예: 모든 "사람" 픽셀을 하나로 묶음)
   - 같은 클래스의 서로 다른 객체를 구분하지 않음

2. **Instance Segmentation (인스턴스 세그멘테이션)** — *YOLO가 주로 지원하는 방식*
   - 같은 클래스라도 개별 객체를 구분해서 각각 마스크 생성
   - 예: 사람이 3명 있으면 각각 다른 마스크로 분리

## YOLO 세그멘테이션 모델 구조

YOLOv8-seg, YOLO11-seg 같은 모델은 다음을 동시에 예측합니다:
- 바운딩 박스 (객체 위치)
- 클래스 (객체 종류)
- **마스크 계수(mask coefficients)** — 프로토타입 마스크와 결합해 최종 픽셀 단위 마스크 생성

내부적으로는 **YOLACT** 방식과 유사하게, 백본에서 나온 "프로토타입 마스크(prototype masks)"와 각 객체별 "마스크 계수"를 선형 결합하여 최종 세그멘테이션 마스크를 만듭니다.


## 활용 분야
- 의료 영상(장기, 종양 경계 추출)
- 자율주행(도로, 차선, 보행자 정밀 구분)
- 농업(작물/잡초 구분)
- 제조업 결함 검사

# YOLO에서의 Pose(포즈 추정) 기능

YOLO의 Pose 기능은 **키포인트 탐지(Keypoint Detection)** 를 통해 사람이나 객체의 **관절 위치와 골격 구조** 를 인식하는 기능입니다. 정식 명칭은 **Pose Estimation(자세 추정)** 입니다.

## 핵심 개념

일반 객체 탐지가 "사람이 여기 있다"(바운딩 박스)를 알려준다면, Pose 기능은 한 단계 더 나아가 **"이 사람의 팔, 다리, 머리 관절이 각각 어디에 있는지"** 까지 좌표로 알려줍니다.

| 구분 | 출력 |
|---|---|
| Detection | 바운딩 박스 |
| Segmentation | 픽셀 단위 마스크 |
| **Pose** | 관절(키포인트) 좌표 + 골격 연결선 |

## 키포인트 구조

YOLO Pose는 일반적으로 **COCO 데이터셋 기준 17개 키포인트**를 사용합니다:

- 코, 좌/우 눈, 좌/우 귀 (얼굴 5개)
- 좌/우 어깨, 좌/우 팔꿈치, 좌/우 손목 (상체 6개)
- 좌/우 골반, 좌/우 무릎, 좌/우 발목 (하체 6개)

각 키포인트는 `(x, y, confidence)` 형태로 예측되며, 이 점들을 이어서 사람 형태의 스켈레톤(뼈대)을 그릴 수 있습니다.

## 출력 방식

각 사람 인스턴스마다:
1. 바운딩 박스 (해당 사람의 위치)
2. 클래스 (사람)
3. 17개 키포인트 좌표 + 각 키포인트의 신뢰도(confidence)

를 동시에 예측합니다.


## 활용 분야
- **피트니스/운동 자세 분석**: 스쿼트, 요가 자세 교정
- **스포츠 분석**: 선수 동작 분석
- **행동 인식**: 낙상 감지, 이상 행동 탐지
- **가상 피팅/AR**: 아바타에 실제 동작 반영
- **재활 치료**: 관절 가동 범위 측정

## 참고
사람 외에도 커스텀 데이터셋으로 학습하면 동물이나 다른 객체의 관절 구조도 추정 가능합니다(예: 동물 행동 분석).

# Pose 결과

## Pose Estimation `Results` 객체 설명

YOLO11 Pose 모델의 `results`는 `ultralytics.engine.results.Results` 객체의 리스트를 반환하며, 각 `Results` 객체는 이미지에서 탐지된 사람들의 포즈 정보(키포인트)를 포함합니다.

주요 속성은 다음과 같습니다:

*   **`boxes`**: 탐지된 사람들의 바운딩 박스 정보 (좌표, 너비, 높이 등)와 클래스 예측, 신뢰도를 포함합니다.
*   **`keypoints`**: 각 탐지된 사람의 키포인트 정보 (예: 어깨, 팔꿈치, 손목 등 관절의 2D 또는 3D 좌표, 신뢰도)를 포함합니다. `result.keypoints`는 `ultralytics.engine.results.Keypoints` 객체이며, `data` 속성에 실제 키포인트 데이터가 Torch 텐서로 저장됩니다.
    *   **`xy`**: 키포인트의 `[x, y]` 좌표를 포함하는 Torch 텐서입니다.
    *   **`conf`**: 각 키포인트의 신뢰도(confidence score)를 포함하는 Torch 텐서입니다.
*   **`names`**: 모델이 탐지할 수 있는 클래스 이름(예: 'person')이 매핑된 딕셔너리입니다.

`result.keypoints.data`는 일반적으로 `[num_instances, num_keypoints, 3]` 형태의 텐서입니다. 여기서 `3`은 `[x_coordinate, y_coordinate, confidence_score]`를 의미합니다.

```python
import torch

# 첫 번째 이미지에 대한 결과 가져오기
result = results[0]

print(f"Results object type: {type(result)}")

# keypoints 정보가 있는지 확인
if result.keypoints is not None:
  print(type(result.keypoints))  # Keypoints 객체
  print(result.keypoints.data.shape)  # [1, 17, 3] -> [검출객체, 키포인트, [x, y, conf]]
  print(result.keypoints.data)

  if len(result.keypoints.data) > 0:
    # 검출된 첫번째 사람의 키포인트만 출력
    first_person_keypoints = result.keypoints.data[0]
    print(first_person_keypoints.cpu().numpy())
```

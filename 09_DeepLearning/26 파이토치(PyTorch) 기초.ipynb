  * 새로운 아이디어를 구현·검증하는 속도가 빠름 → 연구 결과가 곧 PyTorch 생태계로 들어옴

* 산업 적용과 배포도 개선됨
  * 예전에는 TF → 산업, PyTorch → 연구로 나뉘었지만, 현재는 TorchServe, ONNX, PyTorch/XLA, TorchScript로 프로덕션 적용이 쉬워짐  (사실 배포, 모바일 부분은 여전히 TF가 PyTorch에 비해 강점임)
  * Meta, Microsoft, OpenAI 등 주요 빅테크가 PyTorch 기반
  * Hugging Face, OpenAI API 모델 등 최신 LLM/딥러닝 라이브러리도 PyTorch 우선 지원

* API 일관성과 간결함
  * TensorFlow는 tf.keras, tf.data, tf.function 등 하위 모듈 간 사용 방식이 달라서 학습 곡선이 큼
  * PyTorch는 Pythonic한 구조와 명확한 모듈 (torch, torch.nn, torch.utils.data)로 직관적
  * NumPy와 유사한 연산 방식 → 전환 장벽이 낮음

* 생태계 속도와 커뮤니티 대응
  * PyTorch는 새로운 기능 출시와 버그 픽스 속도가 빠름

#### deterministic, non-deterministic 결정적, 비결정적
PyTorch 에서 **비결정적 연산(nondeterministic operations)** 이란, **같은 입력과 같은 조건을 주더라도 실행할 때마다 결과가 달라질 수 있는 연산**을 말합니다. 즉, 완전한 재현성이 보장되지 않는 연산입니다.

---

### 왜 비결정성이 생기는가?

비결정적 연산은 주로 **GPU 병렬 처리와 연관**이 있습니다. GPU에서는 성능 최적화를 위해 여러 쓰레드가 동시에 연산을 수행하는데, 연산 순서가 매번 다를 수 있습니다. 이때 **부동소수점 연산의 누적 오차**나 **atomic 연산의 순서 불확실성** 때문에 결과가 달라지게 됩니다.

---

### 대표적인 비결정적 연산 사례

PyTorch 공식 문서에 따르면, 아래와 같은 연산들이 비결정적일 수 있습니다:

1. **`torch.nn.Conv2d` (특정 cudnn 알고리즘 사용 시)**

   * cuDNN이 내부적으로 빠른 알고리즘을 사용하면서 결과 순서가 달라질 수 있음.

2. **Dropout 같은 난수 기반 연산**

   * 시드를 고정하지 않으면 실행할 때마다 다른 결과가 나옴.
   * 시드를 고정해도 GPU 비결정성 때문에 약간의 차이가 있을 수 있음.

3. **`torch.index_add_`, `scatter_` 같은 in-place 연산**

   * 여러 쓰레드가 같은 메모리 위치를 동시에 업데이트하면 순서에 따라 값이 달라짐.

4. **GPU에서의 backward 연산**

   * 예를 들어, 일부 RNN backward, atomicAdd 기반 gradient accumulation 등이 실행 순서에 따라 달라질 수 있음.

---

### 해결 방법

PyTorch는 **결정성 모드(deterministic mode)** 를 제공하여 가능한 경우 강제로 결정적 연산만 사용하도록 할 수 있습니다:

```python
import torch

# 결정성 강제
torch.use_deterministic_algorithms(True)

# CUDA 환경에서 비결정적 연산을 감지하도록 설정 (디버깅용)
torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False
```

다만, 이 모드를 켜면:

* 일부 연산이 실행 불가(`RuntimeError`)가 될 수 있음
* 성능이 **크게 느려질 수 있음** (특히 GPU에서)

---

✅ 정리하면:

* PyTorch의 **비결정적 연산**은 **병렬 연산 순서 불확실성**이나 **난수 의존성** 때문에 결과가 매번 달라질 수 있는 연산을 말합니다.
* 완전한 재현성이 필요하다면 `torch.use_deterministic_algorithms(True)` 같은 설정으로 **결정성 강제**를 할 수 있지만, **성능 저하**를 감수해야 합니다.



````
"""
※ 참고 램덤시드 주는 방법 정리

import torch
import random
import numpy as np

# Python 내장 random 모듈 시드
random.seed(42)

# NumPy 시드
np.random.seed(42)

# PyTorch 시드
torch.manual_seed(42)

# GPU 사용 시(단일 GPU)
torch.cuda.manual_seed(42)

# 멀티 GPU 환경일 경우
torch.cuda.manual_seed_all(42)

# 완전한 재현성 모드 설정
torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False

실험을 여러번 했을 때 똑같은 결과를 보장하는 것을 '재현성(reproducibility)'  이라 한다

하지만 PyTorch의 완전한 재현성은
동일한 환경(OS, 라이브러리 버전, GPU 드라이버, PyTorch 버전 등)에서만 100% 보장됩니다.
CUDA 연산 중 일부는 여전히 비결정적(non-deterministic)일 수 있어,
필요하면 아래처럼 비결정적 연산 예외 처리도 추가할 수 있습니다.

torch.use_deterministic_algorithms(True)  <- ★주의: jupyter 환경에서 한번 설정하여 이후 셀에서도 계속 정용된다.
"""
None
````

````
#  PyTorch 완전 재현성 보장용 시드 세팅 함수

import os
import random
import numpy as np
import torch

def set_seed(seed: int = 42):
    """
    PyTorch 완전 재현성 보장용 시드 세팅 함수

    Args:
        seed (int): 고정할 시드 값 (기본: 42)
    """
    # Python built-in random
    random.seed(seed)

    # 환경 변수
    os.environ["PYTHONHASHSEED"] = str(seed)

    # NumPy
    np.random.seed(seed)

    # PyTorch CPU
    torch.manual_seed(seed)

    # GPU (단일 및 멀티)
    torch.cuda.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

    # CuDNN 설정 - 완전 결정적 모드
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False

    # PyTorch 연산 결정적 모드 (1.8+)
    torch.use_deterministic_algorithms(True)

    print(f"[INFO] Random seed set to {seed} (full reproducibility mode)")

# 사용 예시
# set_seed(1234)
````

- `.to` 메소드는 Tensor나 nn.Module(모델)을 지정한 장치(device)나 자료형(dtype)으로 변환하는 역할을 합니다.
  - 즉, 연산이 이루어질 **위치(CPU, GPU)**와 **자료형(float32, float16 등)**를 바꾸는 기능

````
"""
※colab 의 경우 런타임을 GPU 로 바꾸고 실행해봅니다
"""
None
````

````
torch.cuda.is_available()  # 확인
````

````
x = torch.randn(1)
print(x)
print(x.item())
print(x.dtype)
````

````
# torch.device('cpu')
# torch.device('cuda')

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(device)  # 'cpu' 혹은 'cuda'


# gpu 가 없는 경우 자동으로 cpu 연결될거다.

# "cpu"
# "device(type='cuda')"
````

````
# device= 매개변수로 어느 device 에 tensor 생성할지 결정
y = torch.ones_like(x, device=device)
y

# tensor([1.])   <- cpu 인 경우
# tensor([1.], device='cuda:0')  <- gpu 인 경우
````

````
# 다른 장치에 생긴 tensor 끼리 연산 불가
# RuntimeError: Expected all tensors to be on the same device, but found at least two devices, cuda:0 and cpu!

# x + y
````

````
x = x.to(device)  # x 를 device 로 옮김
print(x)

# tensor([-0.7245])
# tensor([-0.7245], device='cuda:0')
````

````
# 계산결과도 동일 device 에 생성
z = x + y
print(z)
print(z.to('cpu', torch.double))   # z 를 cpu 로 옮기기

# tensor([0.2755], device='cuda:0')
# tensor([0.2755], dtype=torch.float64)
````

### permute(dims) : 차원순서 재배치

- PyTorch 텐서의 차원 순서를 재배치하는 함수.
원래 데이터의 값은 그대로 두고, 차원 순서만 변경

    - dims: 원하는 차원 순서를 나타내는 튜플 또는 리스트

    - 반환: 새로운 차원 순서를 가진 view (메모리 복사 없이 차원 순서만 변경)

    - 원본 텐서는 변경되지 않음 (필요하면 .clone()으로 복사 가능)


- PyTorch는 이미지 데이터를 다룰때 채널 우선 형식 (C, H, W)을 사용하지만,
PIL 이미지나 NumPy 배열은 채널 마지막 형식 (H, W, C)을 가집니다.

  - 예를 들어, (H=224, W=224, C=3) → (C=3, H=224, W=224) 로 바꾸려면:
  - tensor.permute(2, 0, 1)  # C가 마지막(2) → 앞으로, H=0, W=1

````
# 어떤이미지 데이터: 높이 2, 너비 3, 채널 4
img = torch.randn(2, 3, 4)  # (H, W, C)
print(img, "원래 shape:", img.shape)  # torch.Size([2, 3, 4])
````

````
# 채널 우선 형식 (C, H, W)로 변환
img_ch_first = img.permute(2, 0, 1)
print(img_ch_first, "변경된 shape:", img_ch_first.shape)  # torch.Size([4, 2, 3])
````

### `chunk` 나누기 (개수)
- 텐서를 여러 개로 나눌 때 사용 (몇 개로 나눌 것인가?)

https://docs.pytorch.org/docs/stable/generated/torch.chunk.html

```python
torch.chunk(input: Tensor, chunks: int, dim: int = 0) → Tuple[Tensor, ...]
```

````
x = torch.arange(1, 19).reshape(3, 6)
x
````

````
# x 를 3개로 나눈다, dim=1 차원축을 따라
t1, t2, t3 = torch.chunk(x, 3, dim=1)

# (3,6) => (3,2) , (3,2), (3,2)  3개로 쪼개진다

print(t1)
print(t2)
print(t3)
````

````
t1, t2, t3 = torch.chunk(x, 3, dim=0) # dim=0 차원축을 따라 3개로 나누기

# (3,6) => (1,6), (1,6), (1,6)  3개로 쪼개진다

print(t1)
print(t2)
print(t3)
````

````
torch.chunk(x, 2, dim=0)  # 차원 size 보다 더 적개 쪼개면?  # (3,6) => (2,6) (1,6)
````

````
torch.chunk(x, 4, dim=0)  # 차원 size 보다 더 많이 쪼개면? # (3,6) => (1,6) (1,6) (1,6)
````

# 🟦Autograd(자동미분)

- `torch.autograd` 패키지는 Tensor의 모든 연산에 대해 **자동 미분** 제공
  - '기울기' 를 주어 학습이 이루어지게 함
- 이는 코드를 어떻게 작성하여 실행하느냐에 따라 역전파(back propagation)가 정의된다는 뜻
- `backprop`를 위해 미분값을 자동으로 계산

## requires_grad= 속성
`requires_grad` 속성을 `True`로 설정하면, 해당 텐서에서 이루어지는 모든 연산들을 추적하기 시작

기록을 추적하는 것을 중단하게 하려면, `.detach()`를 호출하여 연산기록으로부터 분리

## 기울기(Gradient)

````
# requires_grad=True 로 tensor 생성.
x = torch.ones(3, 3, requires_grad=True)
x
````

````
x.grad_fn
````

````
# 이제 x 의 연산은 gradient 추적이 된다

y = x + 5  # x 에 덧셈 연산한 결과 y
print(y)   # y 에는 grad_fn=<AddBackward0>  <- x로부터 덧셈한 결과라는 기록이 쌓임
````

````
# 위 y 값에 또 연산을 수행해보자
z = y * y   # 연산
out = z.mean() # 또 연산

print(z)  # grad_fn=<MulBackward0>  곱셈연산에 대한 grad_fn
print(out) # grad_fn=<MeanBackward0> 평균연산에 대한 grad_fns
````

### .backward()  와 .grad속성
계산이 완료된 후, `.backward()`를 호출하면 자동으로 역전파 계산이 가능하고, `.grad` 속성에 누적됨

````
x.grad  # .grad : 미분값, 기울기

# 지금은 계산된게 없다.
# 잠시후 backward() 후에 계산된 것을 보게 될거다!
````

````
# 너무나도 편하게 알아서 미분을 수행해준다!
print(out)
out.backward()  # 역전파 계산!
````

### .data 속성
- torch.Tensor 객체에서 autograd(자동 미분) 관련 정보 없이 순수 값만 담은 텐서를 반환.

- 반환된 텐서는 원본 텐서와 메모리를 공유합니다.

- 주의: autograd 그래프를 무시합니다.
→ .data를 사용해 값을 바꾸면, PyTorch의 미분 기록(그래프)에 반영되지 않아서 그래디언트 계산이 꼬일 수 있음.

- PyTorch 개발자 문서에서도 .data를 권장하지 않음:

  - Instead of using .data, use with torch.no_grad(): or tensor.detach() for safe operations.

## transform 정의
- `DataLoader`의 인자로 들어갈 `transform`을 미리 정의할 수 있고, `Compose`를 통해 리스트 안에 순서대로 전처리 진행

- `ToTensor`()를 하는 이유는 `torchvision`이 PIL Image 형태로만 입력을 받기 때문에 데이터 처리를 위해서 Tensor형으로 변환 필요

  1. 자료형 변환

    - 입력: PIL.Image 또는 NumPy ndarray (dtype=uint8, 0~255)

    - 출력: torch.FloatTensor (dtype=float32)

  1. 차원 순서 변경

    - 일반 이미지: (높이, 너비, 채널) → PyTorch 텐서: (채널, 높이, 너비)

    - 예: (H=224, W=224, C=3) → (C=3, H=224, W=224)

  1. 값 범위 스케일링

    - 0 ~ 255 정수 값 → 0 ~ 1 실수 값

    - 즉, pixel_value / 255.0

````
# transform 정의
mnist_transform = transforms.Compose([
    transforms.ToTensor(), # 타입변환, 차원순서 변경, 스케일링 , 스케일링 [0, 1]
    transforms.Normalize(mean=(0.5,), std=(1.0,)) # 전처리 [0, 1] -> [-0.5, 0.5]
          # 평균이 0에 가까워져서 신경망 학습 (ReLU, tanh) 에서 학습효율이 높아짐
          # 요즘 사전학습모델들 이 이미 [-1, 1] 구간으로 학습된것이 많음.

          # channel 이 3개면. mean=[0.4914, 0.4822, 0.4465],
          #                  std=[0.2023, 0.1994, 0.2010]

])

````

`transforms.Normalize(mean=0.5, std=0.5)` 같은 전처리를 많이 보게 되는데 이유는 다음과 같습니다.

---

### 1. 데이터 정규화의 기본 개념

신경망 학습에서 입력 데이터를 **특정 분포(보통 평균 0, 분산 1)** 로 맞추면 학습이 더 안정적이고 빠르게 진행됩니다.

이유는:

* 각 입력 특성(feature)의 스케일이 비슷해져서 **가중치 업데이트가 균형** 있게 일어남
* 활성화 함수(특히 `sigmoid`, `tanh`)의 **비선형 구간**을 잘 활용할 수 있음
* 경사 하강법에서 **기울기 소실/폭발** 위험을 줄여줌

---

### 2. 0.5, 0.5를 쓰는 구체적 이유

일반적으로 이미지 데이터는 `transforms.ToTensor()` 후 `[0,1]` 범위로 바뀝니다.
여기서 `Normalize(mean=0.5, std=0.5)`를 적용하면:

$x' = \frac{x - 0.5}{0.5} = 2x - 1$

즉, 입력 값 범위가:
$[0, 1] \to [-1, 1]$ 으로 변환됩니다.

---

### 3. 왜 [-1, 1]로 맞추는가?

* **대칭적인 분포**: 평균이 0에 가까워져서 신경망(특히 `tanh`, `ReLU` 류 활성화 함수)이 학습하기 쉬워짐
* **빠른 수렴**: 가중치 초기화와 잘 어울려 학습 속도가 빨라짐
* **실무적 편의**: 많은 pretrained 모델(CNN 등)이 이미 [-1,1] 구간을 가정하고 학습되어 있음

---

### 4. 다른 경우

* CIFAR-10, ImageNet 같은 경우는 실제 데이터셋의 **채널별 평균/표준편차**를 구해서 정규화 (`mean=[0.491, 0.482, 0.447], std=[0.247, 0.243, 0.262]` 등)
* 하지만 간단한 실험이나 GAN 학습에서는 `0.5, 0.5`가 가장 흔히 쓰임 → 픽셀을 단순히 `[-1, 1]`로 스케일링하기 위함

---

✅ 정리

`mean=0.5, std=0.5` 전처리를 하는 이유

 **[0,1] 범위의 이미지를 [-1,1]로 스케일링해서 신경망 학습을 안정적이고 효율적으로 만들기 위해서!**

````
y_train = torch.Tensor([3, 4, 5, 6, 7, 8]).view(6, 1)
y_train
````

````
# DataSet 을 상속 받아 CustomDataset 정의
#   __init__. __getitem__, __len__  구현

class CustomDataset(Dataset):

  # 생성자: 입력데이터, 타겟을 저장
  def __init__(self, x_train, y_train):
    self.x_train = x_train
    self.y_train = y_train

  # index 에 해당하는 아이템 리턴 -> (입력데이터, 타겟) 튜플
  def __getitem__(self, index):
    return self.x_train[index], self.y_train[index]

  # 데이터 사이즈 리턴
  def __len__(self):
    return self.x_train.shape[0]

````

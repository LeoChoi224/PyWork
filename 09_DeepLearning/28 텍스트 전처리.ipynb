# ■ 자연어 처리 (NLP)
Natural Language Processing 이란

**자연어 처리(NLP, Natural Language Processing)** 는 컴퓨터가 인간의 언어를 이해하고, 해석하고, 생성하는 기술을 말합니다. NLP는 언어학과 컴퓨터 과학의 융합 분야로, 언어를 처리하고 분석하는 다양한 작업을 포함합니다. 그 목적은 사람과 컴퓨터 간의 상호작용을 원활하게 만들어 주는 것입니다.

자연어 처리의 주요 작업은 다음과 같습니다:

1. **형태소 분석 (Morphological Analysis)**: 문장에서 단어를 나누고, 각 단어의 형태를 분석합니다. 예를 들어, 한국어에서 "사랑한다"라는 단어는 "사랑"과 "한다"로 나누어집니다.

1. **구문 분석 (Syntactic Parsing)**: 문장의 구조를 분석하고, 문법적 관계를 파악합니다. 예를 들어, "나는 사과를 먹는다"라는 문장에서 주어(나), 목적어(사과), 동사(먹는다) 관계를 식별합니다.

1. **의미 분석 (Semantic Analysis)**: 문장이 전달하려는 의미를 추론합니다. 예를 들어, "배가 고프다"는 단순히 '배'라는 단어가 아니라 '배'가 '허기짐'을 의미하는지를 이해합니다.

1. **감정 분석 (Sentiment Analysis)**: 텍스트에 포함된 감정을 분석하여 긍정적, 부정적, 중립적 감정을 분류합니다. 예를 들어, "이 영화는 정말 재미있다"는 긍정적인 감정을, "이 영화는 지루하다"는 부정적인 감정을 나타냅니다.

1. **기계 번역 (Machine Translation)**: 한 언어에서 다른 언어로 번역하는 작업입니다. 예를 들어, 영어에서 한국어로 번역하는 시스템을 개발하는 것입니다.

1. **질의 응답 (Question Answering)**: 사용자가 입력한 질문에 대해 적절한 답을 제공하는 시스템입니다. 예를 들어, "파리의 수도는 어디인가요?"라는 질문에 대해 "파리는 프랑스의 수도입니다."라는 답을 제공하는 것입니다.

1. **텍스트 생성 (Text Generation)**: 주어진 주제나 조건에 맞춰 텍스트를 자동으로 생성하는 작업입니다. 예를 들어, 사용자가 제공한 주제에 대해 자동으로 기사를 작성하는 것이 이에 해당합니다.

자연어 처리는 다양한 분야에서 활용됩니다. 예를 들어, 검색 엔진, 음성 인식 시스템, 챗봇, 자동 번역 시스템 등에서 사용되며, 최근에는 대규모 언어 모델(예: GPT) 등의 발전으로 자연어 처리 기술이 더욱 진화하고 있습니다.

## 🔹단어 토큰화(Word Tokenization)
토큰의 기준을 단어(word)로 하는 경우, 단어 토큰화(word tokenization)라고 합니다. 다만, 여기서 단어(word)는 단어 단위 외에도 단어구, 의미를 갖는 문자열로도 간주되기도 합니다.
예를 들어보겠습니다. 아래의 입력으로부터 구두점(punctuation)과 같은 문자는 제외시키는 간단한 단어 토큰화 작업을 해봅시다. 구두점이란 마침표(.), 컴마(,), 물음표(?), 세미콜론(;), 느낌표(!) 등과 같은 기호를 말합니다.


입력: **Time is an illusion. Lunchtime double so!**


이러한 입력으로부터 구두점을 제외시킨 토큰화 작업의 결과는 다음과 같습니다.


출력 : "Time", "is", "an", "illustion", "Lunchtime", "double", "so"

### re 사용:
- pytorch 는 직접 대응되는 함수가 없지만 이와 같은 기능을 구현하는 방법들은 여럿 있다


````python
import re

def tokenize_sequence(text, filters=r"[!\"#$%&()*+,-./:;<=>?@[\\]^_`{|}~\t\n]", lower=True, split=" "):
    if lower:
        text = text.lower()
    text = re.sub(filters, "", text)
    return [word for word in text.split(split) if word]
````

````python
print(tokenize_sequence(sentence))
````

## 불용어(Stop word) 제거
불필요한 단어 제거

ex)
1. 등장빈도가 적은 단어
1. 길이가 짧은 단어



````python
# 갖고 있는 데이터에서 유의미한 단어 토큰만을 선별하기 위해서는
# 큰 의미가 없는 단어 토큰을 제거해야 한다

# 예를 들면, I, my, me, over, 조사, 접미사 같은 단어들은 문장에서는 자주 등장하지만
# 실제 의미 분석을 하는데는 거의 기여하는 바가 없는 경우가 있습니다.
# 이러한 단어들을 불용어(stopword)라고 하며,
# NLTK에서는 위와 같은 100여개 이상의 영어 단어들을 불용어로 패키지 내에서 미리 정의하고 있습니다.
````

### NLTK에서 불용어 제거하기

````python
example = "Family is not an important thing. It's everything."
stop_words = set(stopwords.words('english'))

word_tokens = word_tokenize(example)

result = []

for word in word_tokens:
  if word not in stop_words:
    result.append(word)


print('불용어 제거 전 :',word_tokens)
print('불용어 제거 후 :',result)
````

````python
# 단어 인덱스 부여

# 일반적으로 단어의 등장 빈도수를 기준으로 정렬한뒤 부여함
````

## 정수 인코딩

### 1) dict 사용한 인코딩

````python
from nltk.corpus import stopwords
````

````python
raw_text = "A barber is a person. a barber is good person. a barber is huge person. he Knew A Secret! The Secret He Kept is huge secret. Huge secret. His barber kept his word. a barber kept his word. His barber kept his secret. But keeping and keeping such a huge secret to himself was driving the barber crazy. the barber went up a huge mountain."
````

````python
sentences = sent_tokenize(raw_text)
sentences
````

````python
# 정제, 정규화, 단어 토큰화, 단어 빈도

vocab = {}  # 각 단어별 빈도수
preprocessed_sentences = []
stop_words = set(stopwords.words('english'))

for sentence in sentences:
  result = []  # 전처리된 문장의 단어들

  # 각 문장마다 단어 토큰화
  tokenized_sentence = word_tokenize(sentence)

  for word in tokenized_sentence:  # 각 토큰마다
    word = word.lower()  # 모든 단어를 소문자화 단어의 개수를 줄인다.
    if word not in stop_words:  # 불용어 제거
      if len(word) > 2:  # 단어 길이가 2이하인 경우에 대하여 추가로 단어 제거
        result.append(word)

        # 빈도수 체크
        if word not in vocab:
          vocab[word] = 0
        vocab[word] += 1

  preprocessed_sentences.append(result)

preprocessed_sentences
````

````python
# 단어 빈도수
vocab
````

````python
vocab['barber']
````

````python
# 빈도수가 높은 순서대로 정렬
vocab_sorted = sorted(vocab.items(), key = lambda x:x[1], reverse=True)
vocab_sorted
````

````python
# 높은 빈도수를 가진 단어에 낮은 정수 인덱스 부여.  정수값은 1부터 부여

word_to_index= {}

idx = 0
for word, freq in vocab_sorted:
  if freq > 1:  # 빈도수가 작은 단어는 정수 인덱싱 제외.
    idx += 1
    word_to_index[word] = idx

word_to_index
````

````python
# 위에서 자주 사용하는 단어 (어휘) 5개까지만 사용하려면

vocab_size = 5

word_to_index = dict(list(word_to_index.items())[:vocab_size])

word_to_index
````

````python
#  'But keeping and keeping such a huge secret to himself was driving the barber crazy.',

#   ↓  정제 & 토큰화

#  ['keeping', 'keeping', 'huge', 'secret', 'driving', 'barber', 'crazy'],

#   ↓  정수인덱싱 하게 되면?

#  [??,  ??,  3,  2,  ??,  1,  ??]

#  ↑ OOV  어휘 사전에 없는 단어들이 존재할것이다!
````

### OOV (Out-Of-Vocabulary)

````python
# OOV 를 word_to_index  에 새롭게 추가
# 단어집합(사전) 에 없는 단어는 'OOV' 로 인덱싱 할거다.

# ['barber', 'person'] -> [1, 5]
# ['barber', 'good', 'person'] -> [1, ??, 5]   <- 단어집합에 없는 단어는 OOV 로 인코딩
````

````python
# OOV 를 사전에 추가
word_to_index['OOV'] = len(word_to_index) + 1

word_to_index
````

````python
# word_to_index 를 사용하여 sentences 의 모든 단어들을 정수로 인코딩

encoded_sentences = []

for sentence in preprocessed_sentences:
  encoded_sentence = []

  for word in sentence:
    try:
      encoded_sentence.append(word_to_index[word])
    except KeyError:
      encoded_sentence.append(word_to_index['OOV'])

  encoded_sentences.append(encoded_sentence)

encoded_sentences
````

````python
# 원본 문장과 인코딩된 문장(정수 시퀀스)을 함 같이 출력해보자
for e, s in zip(encoded_sentences, sentences):
  print(f'{str(e):25} {s}')
````

## 🧩 대표적인 Special Tokens (BERT 기준)

| 토큰                          | 예시 ID (bert-base-uncased 기준) | 의미              | 역할                                |
| --------------------------- | ---------------------------- | --------------- | --------------------------------- |
| `[PAD]`                     | `0`                          | Padding Token   | 짧은 문장 길이 맞추기용                     |
| `[UNK]`                     | `100`                        | Unknown Token   | 어휘(vocab)에 없는 단어                  |
| `[CLS]`                     | `101`                        | Class Token     | 문장 전체 표현 (classification용)        |
| `[SEP]`                     | `102`                        | Separator Token | 문장 구분 (문장 A/B 구분)                 |
| `[MASK]`                    | `103`                        | Mask Token      | Masked Language Modeling에서 가려진 단어 |
| `[unused0]` ~ `[unused999]` | (104~)                       | 미사용 토큰          | 예비 슬롯 (특수 태그용으로 종종 활용)            |

> 💡 **ID 값은 모델마다 다를 수 있습니다!**
> 예: RoBERTa나 DistilBERT는 `[CLS]`가 `<s>`로, `[SEP]`가 `</s>`로 이름이 다릅니다.

## 🔍 각 토큰의 구체적 역할

### 1️⃣ `[CLS]` (Classify)

* 문장 맨 앞에 항상 추가됨
* 문장 전체의 의미를 요약하는 “대표 토큰”
* 분류(Classification) 작업 시 이 토큰의 벡터를 사용

**예시:**

```python
[CLS] I love PyTorch [SEP]
```

---

### 2️⃣ `[SEP]` (Separator)

* 문장 끝 혹은 문장 사이에 추가됨
* 문장 경계를 표시하거나 문장쌍(A/B) 구분용으로 사용

**예시:**

```python
[CLS] I love PyTorch [SEP] It is great [SEP]
```

---

### 3️⃣ `[PAD]`

* 배치 입력 시 문장 길이를 맞추기 위해 뒤쪽에 채워짐
* 모델이 무시하도록 `attention_mask`에서 0으로 표시됨

```python
[CLS] I love PyTorch [SEP] [PAD] [PAD]
```

---

### 4️⃣ `[UNK]`

* 토크나이저의 어휘 사전에 없는 단어가 들어왔을 때 사용

```python
"I love 🦖" → [CLS] I love [UNK] [SEP]
```

---

### 5️⃣ `[MASK]`

* 사전학습(Pretraining) 중 일부 단어를 가려서 모델이 예측하게 만듦
* Masked Language Modeling(MLM)에서 사용


## 단어집합 (vocabulary)

````python
# 원-핫 인코딩을 위해서 먼저 해야할 일은 '단어 집합'을 만드는 일입니다.
# 텍스트의 모든 단어를 '중복을 허용하지 않고' 모아놓으면 이를 단어 집합

# 그리고 이 단어 집합에 고유한 정수를 부여하는 정수 인코딩을 진행합니다.
# 텍스트에 단어가 총 5,000개가 존재한다면, 단어 집합의 크기는 5,000입니다.

# 각 단어에 고유한 정수 인덱싱이 되었다고 할때
# 이 '숫자'로 바뀐 단어들을 '벡터' 로 다루고 싶다면?
````

## 원핫 인코딩(One-hot encoding) 작성(수동)

````python
# 단어집합의 크기를 벡터의 '차원' 으로 하고
# 표현하고 싶은 단어의 인덱스에 '1' 값을 부여, 다른 인덱스에는 '0' 을 부여하여 단어의 벡터 표현
# 이렇게 표현된 벡터를 원-핫 벡터 (one-hot vector)
````

````python
tokens = okt.morphs("나는 자연어 처리를 배운다")
print(tokens)
````

````python
# 각 토큰에 대해서 고유한 정수를 부여합니다.
# 지금은 문장이 짧기 때문에 각 단어의 빈도수를 고려하지 않지만,
# 빈도수 순으로 단어를 정렬하여 정수를 부여하는 경우가 많습니다.

word_to_index = {word: index for index, word in enumerate(tokens)}
print(word_to_index)
````

````python
def one_hot_encoding(word,word_to_index):
  one_hot_vector = [0] * len(word_to_index)  # 단어 집합의 크기를 벡터의 '차원'으로
  index = word_to_index[word]
  one_hot_vector[index] = 1  # 포현하고 싶은 단어의 인데스에 '1'값 부여

  return one_hot_vector
````

````python
# 자연어 처리(Natural Language Processing)분야에서 임베딩(Embedding)은
# 사람이 쓰는 자연어를 기계가 이해할 수 있는 숫자형태인 vector로 바꾼 결과 혹은 그 일련의 과정 전체를 의미한다.
````

## PyTorch의 nn.Embedding()

**torch.nn.Embedding**
https://docs.pytorch.org/docs/stable/generated/torch.nn.Embedding.html

```python
class torch.nn.Embedding(
  num_embeddings,  # ex) 사전의 크기(입력 차원)
  embedding_dim,  # 임베딩 크기(출력 차원)
  padding_idx=None,
  max_norm=None,
  norm_type=2.0,
  scale_grad_by_freq=False,
  sparse=False,
  _weight=None,
  _freeze=False,
  device=None,
  dtype=None)
```



````python
import torch
import torch.nn as nn

# 1️⃣ 단어 사전 크기(vocab_size)와 임베딩 차원 설정
vocab_size = 10    # 단어가 10개 있다고 가정 (정수 0~9)
embedding_dim = 4  # 각 단어를 4차원 벡터로 표현

# 2️⃣ Embedding 레이어 생성
embedding = nn.Embedding(num_embeddings=vocab_size, embedding_dim=embedding_dim)

# 3️⃣ 입력으로 단어 인덱스(정수)를 전달
# 예: 문장 [2, 5, 7] 을 임베딩에 통과
input_indices = torch.tensor([2, 5, 7])  # 정수형 텐서 (예: [2, 5, 7])

# 4️⃣ 임베딩 결과 출력
embedded = embedding(input_indices)

print("입력 인덱스:", input_indices)
print("임베딩 결과 텐서 크기:", embedded.shape)
  # 결과 텐서 크기 [입력길이, embedding_dim]  즉, [3, 4] 형태
print("임베딩 벡터:\n", embedded)

# 출력 예시 (랜덤 초기화됨)
````

````python
# 문장을 여러 단어로 구성된 배치(batch) 로 처리할 수도 있습니다.

# batch_size = 2, 문장 길이 = 3

input_batch = torch.tensor([[1, 2, 3],
                            [4, 5, 6]])

embedded_batch = embedding(input_batch)  # (2, 3) -> (2, 3, 4)

embedded_batch.shape  # 배치 2개, 각 문장(시퀀스) 3단어, 임베딩 차원 4
````

````python
# nn.Embedding 역할
# 정수 인덱스를 실수 벡터로 변환
# 입력  LongTensor (ex: [1, 2, 3])
# 출력  FloatTensor (ex: [3, embedding_dim])
# 학습  다른 신경망 층과 함께 학습 가능 (임베딩 벡터는 requires_grad=True)
````
